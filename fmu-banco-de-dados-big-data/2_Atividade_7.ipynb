```python
#importar o pymongo
from pymongo import MongoClient

URL="SUA_STRING_DE_CONEXAO"
#conexâo
client=MongoClient(URL)

#selecionar o database
db=client["clinica_veterinaria"]

#selecionando a collection
colecao=db["animais"]

print("Conectado ao MongoDB com sucesso!")
```

```python
#Crud - create - insert

animal={
    "nome":"Rex",
    "especie":"Cachorro",
    "idade":5,
    "peso":18.5,
    "vacinado":True,
    "tutor":"Carlos Silva",
    "cidade":"São Paulo"
}

resultado=colecao.insert_one(animal)

print(f"Id_criado: {resultado.inserted_id}")
```

```python
animais=[

    {
        "nome":"Mimi",
        "especie":"Gato",
        "idade":2,
        "peso":4.2,
        "vacinado":True,
        "tutor":"Ana Souza",
        "cidade":"Campinas"
    },

    {
        "nome":"Thor",
        "especie":"Cachorro",
        "idade":7,
        "peso":25,
        "vacinado":False,
        "tutor":"Mariana Lima",
        "cidade":"São Paulo"
    },

    {
        "nome":"Luna",
        "especie":"Coelho",
        "idade":1,
        "peso":2.1,
        "vacinado":True,
        "tutor":"Pedro Alves",
        "cidade":"Santos"
    },

    {
        "nome":"Bob",
        "especie":"Cachorro",
        "idade":3,
        "peso":12,
        "vacinado":True,
        "tutor":"Fernanda Rocha",
        "cidade":"Sorocaba"
    }

]

resultado=colecao.insert_many(animais)

print(f"Foram inseridos {len(resultado.inserted_ids)} animais")
```

```python
#Read - imprimindo os alunos armazenados
#cRud --> R read - ler os doctos que estão gravados
#find() sem parâmetros = traga todos os documentos gravados

for animal in colecao.find():
    print(animal)
```

```python
#Mostrar apenas os cachorros

filtro={"especie":"Cachorro"}

for animal in colecao.find(filtro):
    print(animal)

#Buscar os animais vacinados

filtro={"vacinado":True}

for animal in colecao.find(filtro):
    print(animal)

#Buscar animais com idade maior que 3 anos

filtro={"idade":{"$gt":3}}

for animal in colecao.find(filtro):
    print(animal)

#Listar os animais da cidade de São Paulo

filtro={"cidade":"São Paulo"}

for animal in colecao.find(filtro):
    print(animal)
```

```python
#Update - alterar a idade do Rex para 6

filtro={"nome":"Rex"}
novos_dados={"$set":{"idade":6}}
resultado=colecao.update_one(filtro,novos_dados)

print(f"Nº de documentos atualizados: {resultado.modified_count}")

#Update - alterar vacinado para True de todos os animais

filtro={}

novos_dados={"$set":{"vacinado":True}}

resultado=colecao.update_many(filtro,novos_dados)

print(f"Nº de documentos atualizados: {resultado.modified_count}")
```
