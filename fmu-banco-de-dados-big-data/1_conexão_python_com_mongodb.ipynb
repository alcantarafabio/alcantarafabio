```python
#importar o pymongo
from pymongo import MongoClient
#ROTEIRO
#criar um banco de dados: faculdade
#criar a collection: alunos
#criar um novo usuário fabiouser / 246810 (password)
#criar o ip access list: 0.0.0.0/0 (acessa de qquer lugar)
#conectar-se ao cluster0 através do project overview e
#ecolher mongo com vscode - copiar a string de url
#e substituir pelo user name e senha
#conexão com o banco de dados: faculdade
#usuário: fabiouser
#password: 246810

URL="SUA_STRING_DE_CONEXAO"
#conexâo
client=MongoClient(URL)

#selecionar o database
db=client["faculdade"]

#selecionando a collection
colecao=db["alunos"]

#print("Conectado ao MongoDB com sucesso!")
```

```python
#Crud - create - insert
aluno={
    "nome" : "Maria Silva",
    "idade" : 20,
    "curso" : "Ciência de Dados",
    "semestre" : 2,
    "nota" : [8.5,9.0,7.5]
}

resultado=colecao.insert_one(aluno)
print(f"Id_criado: {resultado.inserted_id}")
```

```python
turma=[
    {"nome":"Cristina", "idade" : 44, "curso" : "Enfermagem"},
    {"nome":"Ricardo", "idade" : 33, "curso" : "Ciência de Dados"},
    {"nome":"José", "idade" : 30, "curso" : "Engenharia da Computação"}
]
resultado=colecao.insert_many(turma)
print(f"Foram inseridos {len(resultado.inserted_ids)} alunos")
```

```python
#Read - imprimindo os alunos armazenados
#cRud --> R read - ler os doctos que estão gravados
#find() sem parâmetros = traga todos os documentos gravados
for docto in colecao.find():
  print(docto)
```

```python
#Leitura com filtro
#filtro={"idade": 20}
filtro={"curso": {"$regex": "^engenharia", "$options": "i"}}
for aluno in colecao.find(filtro):
  print(aluno)
```

```python
#crUd --> U -> update (atualização) os doctos que estão gravados
#semestre da Maria Silva está 2 e queremos que seja 4
# $set garante que o documento será substituido
#CUIDADO!!!!! PRESTE ATENÇÃO AO FILTRO

filtro= {"nome" : "Maria Silva"}
novos_dados={"$set" : {"semestre" : 4}}
resultado=colecao.update_one(filtro,novos_dados)
print(f"Nº de doctos atualizados: {resultado.modified_count}")
```

```python
filtro= {"nome" : "Cristina"}
novos_dados={"$set" : {"idade" :100}}
resultado=colecao.update_many(filtro,novos_dados)
print(f"Nº de doctos atualizados: {resultado.modified_count}")
```

```python
#cruD -->D -> delete - exclui os doctos selecionados
#Excluir o primeiro documentos cujo nome é Pedro
#Atenção: estamos utilizando a variável filtro criada
#na célula anterior
filtro={"nome" : {"$regex": "cRIStina","$options": "i"}}
resultado=colecao.delete_one(filtro)
print(f"Documentos deletados: {resultado.deleted_count}")
```
